```python
import cv2
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
```

## 1. Read and display flower.jpg using OpenCV

```python
img = cv2.imread("flower.jpg")   # Read image using OpenCV
cv2.imshow("Original Image", img) # Display image with title
cv2.waitKey(0)                    # Wait for a key press
cv2.destroyAllWindows()           # Close the window
```

## 2. Print dimensions and data type of scene.jpg

```python
img = cv2.imread("scene.jpg")  # Load the image
print("Shape:", img.shape)     # Print height, width and channels
print("Datatype:", img.dtype)  # Print pixel data type
```

## 3. Convert a colour image to grayscale and save it

```python
img_cv = cv2.imread("scene.jpg")  # Load colour image
gray_cv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)  # Convert to grayscale
cv2.imwrite("output_gray.jpg", gray_cv)  # Save grayscale image
```

## 5. Resize an image to 50%

```python
img_cv = cv2.imread("flower.jpg")  # Load image
resized = cv2.resize(img_cv, None, fx=0.5, fy=0.5)  # Resize to 50%
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))  # Display in RGB
plt.title("Resized Image")  # Add title
plt.axis("off")             # Hide axes
plt.show()                  # Display image
```

## 6. Crop the central 200 × 200 area

```python
img_cv = cv2.imread("flower.jpg")  # Load image
h, w = img_cv.shape[:2]           # Get height and width
x, y = w // 2, h // 2             # Find centre coordinates
cropped = img_cv[y-100:y+100, x-100:x+100]  # Crop central area
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))  # Display crop
plt.title("Cropped Image")        # Add title
plt.axis("off")                   # Hide axes
plt.show()                        # Show image
```

## 7. Rotate an image 90° clockwise

```python
img_cv = cv2.imread("flower.jpg")  # Load image
rotated = cv2.rotate(img_cv, cv2.ROTATE_90_CLOCKWISE)  # Rotate
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))  # Display image
plt.title("Rotated Image")         # Add title
plt.axis("off")                    # Hide axes
plt.show()                         # Show image
```

## 8. Flip car.jpg left-to-right

```python
img_cv = cv2.imread("car.jpg")  # Load car image
flip_horizontal = cv2.flip(img_cv, 1)  # Flip horizontally
plt.imshow(cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB))  # Display
plt.title("Horizontal Flip")    # Add title
plt.axis("off")                 # Hide axes
plt.show()                      # Show image
```

## 9. Convert RGB to HSV and display each channel

```python
img_cv = cv2.imread("flower.jpg")  # Load image in BGR
rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB
hsv = cv2.cvtColor(rgb, cv2.COLOR_RGB2HSV)  # Convert RGB to HSV
h, s, v = cv2.split(hsv)  # Separate H, S and V channels

plt.figure(figsize=(12, 3))  # Set figure size
for i, (img, title) in enumerate(zip([h, s, v], ["Hue", "Saturation", "Value"])):
    plt.subplot(1, 3, i+1)  # Create three plots
    plt.imshow(img, cmap="gray")  # Display channel
    plt.title(title)  # Add channel name
    plt.axis("off")  # Hide axes

plt.show()  # Display all three channels
```

## 10. Calculate MSE between two grayscale images

```python
img1 = cv2.imread("img1.jpg", cv2.IMREAD_GRAYSCALE)  # Load first image
img2 = cv2.imread("img2.jpg", cv2.IMREAD_GRAYSCALE)  # Load second image

difference = img1.astype(np.float64) - img2.astype(np.float64)  # Find difference
mse = np.mean(difference ** 2)  # Calculate average squared difference
print("Mean Squared Error (MSE):", mse)  # Display result
```
